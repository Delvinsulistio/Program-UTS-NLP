# Pembersihan Data Steam Review Game

Notebook ini digunakan untuk membersihkan dataset `steam_review_game.csv` agar hanya menyisakan kolom yang diperlukan (`appid`, `review`, `name`) dan menghapus data-data yang tidak valid.

**Langkah-langkah:**
1. Import library & load data
2. Lihat info awal dataset
3. Hapus kolom yang tidak diperlukan
4. Hapus missing values pada kolom appid, review, dan name
5. Bersihkan whitespace pada kolom review dan name
6. Hapus review yang kosong (empty string / whitespace only)
7. Hapus duplikat berdasarkan kolom review
8. Lihat info dataset setelah dibersihkan
9. Simpan ke file CSV baru

```python
# Import library yang diperlukan dan load dataset
import pandas as pd

df = pd.read_csv('steam_review_game.csv')
print(f'Dataset berhasil dimuat!')
print(f'Jumlah baris: {len(df):,}')
print(f'Jumlah kolom: {len(df.columns)}')
print(f'Kolom: {list(df.columns)}')
```

```python
# Kode ini untuk menampilkan info awal dataset sebelum dibersihkan
print('=== INFO AWAL DATASET ===')
print(f'Shape: {df.shape}')
print(f'\nMissing values per kolom:')
print(df.isnull().sum())
print(f'\nSample data (5 baris pertama):')
df.head()
```

```python
# Kode ini untuk menghapus kolom yang tidak diperlukan
# Hanya menyisakan kolom: appid, review, name
kolom_dihapus = [col for col in df.columns if col not in ['appid', 'review', 'name']]
print(f'Kolom yang akan dihapus: {kolom_dihapus}')

df = df[['appid', 'review', 'name']]

print(f'\nKolom yang tersisa: {list(df.columns)}')
print(f'Shape setelah hapus kolom: {df.shape}')
```

```python
# Kode ini untuk menghapus baris yang memiliki missing value (NaN/null)
# pada kolom appid, review, atau name
jumlah_sebelum = len(df)

print('Missing values sebelum dihapus:')
print(df.isnull().sum())

df = df.dropna(subset=['appid', 'review', 'name'])

jumlah_sesudah = len(df)
print(f'\nBaris dihapus karena missing value: {jumlah_sebelum - jumlah_sesudah:,}')
print(f'Sisa baris: {jumlah_sesudah:,}')
```

```python
# Kode ini untuk membersihkan whitespace (spasi berlebih) di kolom review dan name
# Strip leading/trailing whitespace
df['review'] = df['review'].astype(str).str.strip()
df['name'] = df['name'].astype(str).str.strip()

print('Whitespace pada kolom review dan name sudah dibersihkan.')
```

```python
# Kode ini untuk menghapus baris yang reviewnya kosong (empty string)
# setelah di-strip whitespace-nya
jumlah_sebelum = len(df)

df = df[df['review'] != '']

jumlah_sesudah = len(df)
print(f'Baris dihapus karena review kosong: {jumlah_sebelum - jumlah_sesudah:,}')
print(f'Sisa baris: {jumlah_sesudah:,}')
```

```python
# Kode ini untuk menghapus duplikat berdasarkan kolom review
# Catatan: appid dan name TIDAK dijadikan basis duplikat, karena
# satu game bisa punya banyak review (appid & name pasti duplikat)
# Yang dihapus hanya jika teks review-nya persis sama
jumlah_sebelum = len(df)

df = df.drop_duplicates(subset=['review'], keep='first')

jumlah_sesudah = len(df)
print(f'Baris dihapus karena review duplikat: {jumlah_sebelum - jumlah_sesudah:,}')
print(f'Sisa baris: {jumlah_sesudah:,}')
```

```python
# Kode ini untuk reset index setelah semua pembersihan
df = df.reset_index(drop=True)

print('Index sudah di-reset.')
print(f'Shape final: {df.shape}')
```

```python
# Kode ini untuk menampilkan info dataset setelah dibersihkan
print('=== INFO DATASET SETELAH DIBERSIHKAN ===')
print(f'Jumlah baris: {len(df):,}')
print(f'Jumlah kolom: {len(df.columns)}')
print(f'Kolom: {list(df.columns)}')
print(f'\nMissing values:')
print(df.isnull().sum())
print(f'\nDuplikat review: {df.duplicated(subset=["review"]).sum()}')
print(f'\nSample data (5 baris pertama):')
df.head()
```

```python
# Kode ini untuk menyimpan dataset yang sudah bersih ke file CSV baru
df.to_csv('steam_review_game_bersih.csv', index=False)

print(f'Dataset bersih berhasil disimpan ke: steam_review_game_bersih.csv')
print(f'Total baris yang tersimpan: {len(df):,}')
```
